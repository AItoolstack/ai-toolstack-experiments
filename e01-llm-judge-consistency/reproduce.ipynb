## 4. Why did the judges disagree? Read their reasons

```python
import json
from collections import Counter
from common import JUDGES, load_runs

ITEM = 'f07'  # change me
runs = load_runs('items_v2.jsonl')
for stem, name in JUDGES.items():
    rows = [r for r in runs[stem] if r['item'] == ITEM]
    print(name, dict(Counter(r['verdict'] for r in rows)))
    for r in rows[:2]:
        print('   ', r['verdict'], r['raw'][:200])
```
